## Extracting Greek Sections:

```python
with open("../01_topics_aristotle/aristotle_topics_greek.txt", 'r') as f:
    grc_text = f.read()

grc_text[:100]

pattern = r"\[\d+\.\d+\]"

re.findall(pattern, grc_text)
```

```python
grc_sections = re.split(pattern, grc_text)

print(len(grc_sections))
display(grc_sections)
```

```python
import unicodedata

def normalize_greek(text, strip_diacritics=False, normalize_sigma=False):
    """
    Normalize Ancient Greek text.

    Args:
        text (str): Greek text.
        strip_diacritics (bool): Remove accents, breathings, etc.
        normalize_sigma (bool): convert final sigma (ς) to the standard sigma (σ)

    Returns:
        str: Normalized text.
    """
    # Canonical Unicode normalization
    text = unicodedata.normalize("NFC", text)

    # I'll keep capitalization for now.
    # text = text.lower()

    GREEK_RE = re.compile(r"[^Ͱ-Ͽἀ-῿\s]+") 
    text = GREEK_RE.sub("", text)      # Remove non-Greek characters
    text = re.sub(r"\s+", " ", text).strip()

    if strip_diacritics:
        # Decompose characters
        text = unicodedata.normalize("NFD", text)
        # Remove combining marks
        text = "".join(
            c for c in text
            if unicodedata.category(c) != "Mn"
        )
        # Recompose
        text = unicodedata.normalize("NFC", text)

    if normalize_sigma:
        text = text.replace("ς", "σ")

    return text



grc_sections_clean = []
for sec in grc_sections[1:]: 
    grc_sections_clean.append(normalize_greek(sec))
```

```python
print(len(grc_sections_clean))
display(grc_sections_clean)
```
